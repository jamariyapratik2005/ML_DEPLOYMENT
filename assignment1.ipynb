## Question 1
**Write a short paragraph explaining the difference between training and deployment of a machine learning model, using an example from an app you use daily (such as Instagram's feed ranking or Zomato's food recommendations).**

### Answer:

**Training** is the phase where a machine learning model *learns patterns* from historical data, while **Deployment** is the phase where that trained model is *put into production* to make predictions on new, unseen data in real time.

**Example — Instagram's Feed Ranking:**

During **training**, Instagram collects billions of data points — which posts you liked, commented on, shared, how long you paused on a reel, whom you interact with most — and feeds this data into a ranking model so it can learn what kind of content each user prefers. This training happens on powerful servers using historical interaction data. During **deployment**, the trained model runs live on Instagram's servers. Every time you open the app and pull down to refresh, the deployed model scores and ranks hundreds of candidate posts in milliseconds, deciding the exact order in which posts appear in *your* feed. So, training is the "learning" phase (offline, on past data), while deployment is the "serving" phase (online, on live user requests).

## Question 2
**List three real-world examples where ML models are deployed in popular apps (for example, spam detection in WhatsApp, personalized playlists in Spotify, or fraud detection in Paytm), and for each, briefly describe how users interact with the ML-powered feature.**

### Answer:

**1. Spam Detection in Gmail / WhatsApp:**
Every time a user receives an email or message, a deployed ML model silently analyzes the content, sender reputation, and patterns in the background. If the model classifies the message as spam, it is automatically moved to the spam folder or flagged with a warning — the user simply sees a clean inbox without having to manually filter junk.

**2. Personalized Playlists in Spotify (e.g., Discover Weekly):**
When a user opens Spotify on Monday, they find a fresh playlist of 30 songs tailored specifically to their taste. Behind the scenes, a deployed recommendation model analyzed their listening history, skipped tracks, liked songs, and similar users' preferences to curate this playlist — the user just hits play and enjoys music they've never heard before.

**3. Fraud Detection in Paytm / Google Pay:**
When a user initiates a UPI transaction, a deployed fraud-detection model evaluates the transaction in real time — checking the amount, recipient, device, location, and behavioral patterns. If the model flags it as suspicious, the transaction is either blocked or the user is prompted with an extra verification step (like OTP or biometric), protecting them from unauthorized payments.

## Question 3
**Imagine you are building a movie recommendation feature for a BookMyShow-like app. Describe whether you would use online or offline deployment for your ML model and justify your choice in 3-4 sentences.**

*Hint: Consider how quickly recommendations need to update and how often the model is retrained.*

### Answer:

I would use **offline (batch) deployment** for the movie recommendation model. The movie catalog on a BookMyShow-like app doesn't change every second — new movies release weekly, and a user's broad taste in genres evolves slowly over time, not between every click. Batch processing allows us to pre-compute personalized recommendation lists for all users at scheduled intervals (e.g., every few hours or once a day) efficiently, without worrying about real-time latency. Retraining the model can happen weekly or bi-weekly since user preferences and the movie catalog don't shift drastically day-to-day, making offline deployment the most cost-effective and practical choice.

## Question 4
**Given a list of 10 user reviews for a restaurant (as text), write a Python function that simulates batch prediction by categorizing each review as 'positive' or 'negative' based on whether it contains the word 'good'. Print the results for all reviews.**

*Constraint: Do not use any external ML libraries; just use basic string operations.*

```python
def batch_predict_reviews(reviews):
    """
    Simulates batch prediction by categorizing each review as
    'positive' or 'negative' based on whether it contains the word 'good'.
    """
    results = []
    for review in reviews:
        # Case-insensitive check for the word 'good'
        if "good" in review.lower():
            sentiment = "positive"
        else:
            sentiment = "negative"
        results.append({"review": review, "prediction": sentiment})
    return results


# --- Sample Data: 10 restaurant reviews ---
reviews = [
    "The food was really good and fresh!",
    "Terrible service, waited 45 minutes for cold food.",
    "Good ambiance and friendly staff.",
    "The worst dining experience I've ever had.",
    "Pretty good pizza, would visit again.",
    "Overpriced and tasteless, very disappointing.",
    "Had a good time with family, loved the desserts.",
    "Food was stale and the place was dirty.",
    "Not bad, the biryani was surprisingly good!",
    "Rude staff and horrible food, never coming back.",
]

# --- Run Batch Prediction ---
predictions = batch_predict_reviews(reviews)

# --- Print Results ---
print(f"{'No.':<5} {'Prediction':<12} {'Review'}")
print("-" * 70)
for i, result in enumerate(predictions, start=1):
    print(f"{i:<5} {result['prediction']:<12} {result['review']}")
```

## Question 5
**Think of a common deployment challenge (such as model version mismatch, slow predictions, or scaling to many users) that could happen if you were deploying an ML model in a Flipkart-like app. Briefly describe the challenge and suggest one solution to handle it.**

### Answer:

**Challenge: Scaling to Millions of Concurrent Users**

In a Flipkart-like e-commerce app, especially during flash sales (like *Big Billion Days*), the number of users hitting the recommendation and search-ranking ML models can spike from a few lakhs to several crores of requests per minute. If the deployed model infrastructure isn't designed for this scale, users will experience slow page loads, delayed recommendations, or outright failures — leading to a poor shopping experience and lost revenue.

**Solution: Auto-Scaling with Model Serving Infrastructure**

Deploy the ML model behind an auto-scaling serving infrastructure (e.g., Kubernetes with Horizontal Pod Autoscaler, or managed services like AWS SageMaker). This automatically spins up more model-serving instances when traffic spikes and scales down during low-traffic periods, ensuring consistent low-latency predictions. Additionally, caching frequently requested predictions in a fast in-memory store like Redis can reduce redundant model inference and dramatically lower the load on model servers during peak traffic.
